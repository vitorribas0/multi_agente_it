# Teste Vertex (IaraGenAI)

Notebook simples para validar o provider `vertex` via `IaraGenAI`, usando a tool `enterpriseWebSearch`.

> **SI:** a tool `googleSearch` **não** é mais recomendada (não aderente aos requisitos de SI). Use `enterpriseWebSearch`.
>
> Parecer de SI: https://iconectados.sharepoint.com/sites/AIRTCyber/SitePages/GCP-VertexAI---Gemini.aspx

```python
import os
from uuid import uuid4

from dotenv import load_dotenv
from iaragenai import IaraGenAI

load_dotenv()
```

```python
client = IaraGenAI(
    client_id=os.getenv("IARA_CLIENT_ID", "your_client_id"),
    client_secret=os.getenv("IARA_CLIENT_SECRET", "your_client_secret"),
    environment=os.getenv("IARA_ENVIRONMENT", "dev"),
    provider="vertex",
)
```

```python
# ATENÇÃO: o uso da tool googleSearch não é mais recomendado (não aderente aos requisitos de SI).
# Use a tool enterpriseWebSearch.
tools = [{"enterpriseWebSearch": {}}]

chat_completion = client.with_options(
    correlation_id=str(uuid4())
).chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "me traga noticias do dia 21/07/2026 ou 20/07/2026, sem mentir, me traga somente noticias desse dia, se nâo tiver me avise, caso encontre, me traga o link da noticias, e onde você encontrou essa noticia" ,
        }
    ],
    tools=tools,
    model="gemini-2.5-flash-lite",
    max_tokens=20000,
    temperature=0.9,
    top_p=0.95,
)
```

```python
print(chat_completion.choices[0].message.content)
```
